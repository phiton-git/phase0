## 3. Data Loading

### 3.1 Import libraries

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
sns.set_theme(style="whitegrid")

DATA_PATH = Path("P0M1_nawafi_suren_dataset.csv")
df_raw = pd.read_csv(DATA_PATH, parse_dates=["listing_date"])

print("Raw shape:", df_raw.shape)
display(df_raw.head())
````

### 3.2 Eksplorasi awal

Periksa tipe data, statistik awal, missing values, duplikasi, dan rentang nilai. Tahap ini membantu menemukan masalah sebelum analisis.

````python
display(df_raw.info())
display(df_raw.describe(include="all").T)
print("Missing values per column:")
display(df_raw.isna().sum())
print("Duplicate rows:", df_raw.duplicated().sum())
````

## 4. Data Cleaning

Strategi cleaning:
1. Salin data mentah agar sumber awal tetap tersedia.
2. Hapus duplikasi berdasarkan `listing_id` (satu ID mewakili satu listing).
3. Ubah tipe data tanggal.
4. Hapus baris dengan nilai tidak masuk akal: harga <= 0, luas <= 0, atau jumlah kamar negatif.
5. Isi missing numeric dengan median agar baris lain tetap dapat digunakan.
6. Validasi hubungan luas bangunan tidak melebihi luas tanah pada simulasi ini.
7. Buat kolom turunan `price_million` untuk memudahkan pembacaan harga dalam juta rupiah.

Catatan: pada data nyata, keputusan imputasi dan penghapusan harus didasarkan pada konteks bisnis, bukan dilakukan otomatis tanpa pemeriksaan.

````python
df = df_raw.copy()

# 1. Remove duplicated listing IDs
df = df.drop_duplicates(subset="listing_id", keep="first").copy()

# 2. Ensure correct types
df["listing_date"] = pd.to_datetime(df["listing_date"], errors="coerce")
numeric_cols = [
    "land_area_m2", "building_area_m2", "bedrooms", "bathrooms",
    "carport", "renovated", "property_age_years", "price_idr",
    "latitude", "longitude"
]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 3. Remove invalid values
df = df[
    (df["price_idr"] > 0) &
    (df["land_area_m2"] > 0) &
    (df["building_area_m2"] > 0) &
    (df["bedrooms"] >= 0) &
    (df["bathrooms"] >= 0) &
    (df["carport"] >= 0)
].copy()

# 4. Impute missing numeric values with median
for col in numeric_cols:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].median())

# 5. Keep physically plausible records for this simulated dataset
df = df[df["building_area_m2"] <= df["land_area_m2"]].copy()

# 6. Derived columns
df["price_million"] = df["price_idr"] / 1_000_000
df["carport_group"] = np.where(df["carport"] > 0, "With carport", "No carport")
df["listing_month"] = df["listing_date"].dt.month_name()

print("Clean shape:", df.shape)
print("Remaining missing values:", int(df.isna().sum().sum()))
print("Remaining duplicate IDs:", df["listing_id"].duplicated().sum())
display(df.head())
````

### 4.1 Cleaning summary

Bandingkan ukuran data sebelum dan sesudah cleaning. Perubahan jumlah baris terjadi karena duplikasi dan nilai tidak valid yang sengaja dimasukkan untuk latihan.

````python
cleaning_summary = pd.DataFrame({
    "stage": ["Raw data", "After cleaning"],
    "rows": [len(df_raw), len(df)]
})
display(cleaning_summary)
````

### 5.5 Statistik deskriptif – Harga properti

Gunakan mean, median, standard deviation, variance, quartile, IQR, dan aturan 1.5 × IQR untuk menandai potensi outlier.

Outlier statistik tidak otomatis berarti data salah. Listing mahal bisa jadi properti premium yang valid.

````python
price = df["price_million"]
desc = price.describe()
q1 = price.quantile(0.25)
q3 = price.quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = df[(price < lower_bound) | (price > upper_bound)]

descriptive_summary = pd.DataFrame({
    "metric": ["Mean", "Median", "Mode", "Variance", "Std deviation",
               "Minimum", "Q1", "Q3", "Maximum", "IQR",
               "Lower outlier bound", "Upper outlier bound", "Outlier count"],
    "value": [
        price.mean(), price.median(), price.mode().iloc[0],
        price.var(), price.std(), price.min(), q1, q3, price.max(),
        iqr, lower_bound, upper_bound, len(outliers)
    ]
})
display(descriptive_summary)
````

Interpretasi:
- Mean adalah rata-rata harga.
- Median adalah nilai tengah setelah data diurutkan.
- Standard deviation menggambarkan seberapa menyebar harga dari mean.
- IQR adalah Q3 − Q1, yaitu rentang 50% data tengah.
- Batas outlier dihitung dengan Q1 − 1,5×IQR dan Q3 + 1,5×IQR.

Gunakan hasil angka yang muncul ketika notebook dijalankan untuk menulis interpretasi final.

### 5.6 Statistik inferensial – Apakah harga berbeda berdasarkan carport?

**Pertanyaan:** apakah rata-rata harga penawaran rumah dengan carport berbeda dari rumah tanpa carport?

- H0: rata-rata harga kedua kelompok sama.
- H1: rata-rata harga kedua kelompok berbeda.
- Alpha: 0,05.
- Uji: Welch independent two-sample t-test, tidak mengasumsikan varians kedua kelompok sama.
- Kelompok: `carport > 0` dan `carport = 0`.

Uji ini menguji perbedaan rata-rata pada dataset simulasi. Karena data bukan sampel acak dari transaksi pasar nyata, hasilnya tidak boleh digeneralisasi ke seluruh pasar Bekasi.

````python
with_carport = df.loc[df["carport"] > 0, "price_million"]
without_carport = df.loc[df["carport"] == 0, "price_million"]

t_stat, p_value = stats.ttest_ind(
    with_carport, without_carport,
    equal_var=False,
    alternative="two-sided"
)

mean_with = with_carport.mean()
mean_without = without_carport.mean()
difference = mean_with - mean_without

# Welch confidence interval for difference in means
n1, n0 = len(with_carport), len(without_carport)
v1, v0 = with_carport.var(ddof=1), without_carport.var(ddof=1)
se = np.sqrt(v1/n1 + v0/n0)
df_welch = (v1/n1 + v0/n0)**2 / (
    (v1/n1)**2/(n1-1) + (v0/n0)**2/(n0-1)
)
critical_t = stats.t.ppf(0.975, df_welch)
ci_low = difference - critical_t * se
ci_high = difference + critical_t * se

print(f"n with carport: {n1:,}")
print(f"n without carport: {n0:,}")
print(f"Mean with carport: {mean_with:,.2f} million IDR")
print(f"Mean without carport: {mean_without:,.2f} million IDR")
print(f"Mean difference: {difference:,.2f} million IDR")
print(f"95% CI difference: [{ci_low:,.2f}, {ci_high:,.2f}]")
print(f"Welch t-statistic: {t_stat:.3f}")
print(f"p-value: {p_value:.6f}")

if p_value < 0.05:
    print("Keputusan: Tolak H0. Terdapat bukti perbedaan rata-rata harga pada data simulasi.")
else:
    print("Keputusan: Gagal menolak H0. Belum cukup bukti perbedaan rata-rata harga.")
````

Interpretasi yang wajib ditulis:
1. Bandingkan mean kelompok dengan dan tanpa carport.
2. Periksa p-value terhadap alpha 0,05.
3. Periksa apakah confidence interval selisih mean mencakup nol.
4. Hindari menyatakan carport menyebabkan harga naik. Uji ini hanya menunjukkan asosiasi/perbedaan kelompok, bukan kausalitas.
